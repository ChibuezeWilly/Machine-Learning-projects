User Behavior Clustering (App Analytics)
🎯 Goal:

Group users based on how they use an app

```python
from sklearn._config import set_config
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
import pandas as pd
import matplotlib.pyplot as plt
```

```python
set_config(transform_output='pandas')

# load dataset
df = pd.read_csv('behaviour.csv')
X = df
# scale features
scaled_data = StandardScaler().fit_transform(X)

# create clusters.
model = KMeans(n_clusters=3, init='k-means++', random_state=42, n_init=10)
df['Cluster'] = model.fit_predict(scaled_data)
labels = df['Cluster']

sil_score = silhouette_score(scaled_data, labels)  
print(f"Silhouette score: {sil_score}")
```

my sil score was 0.57 which is a very good score because anythng from 0.70 means the data is fake or too simple.

And from the dataset, we have 3 types of users.
User 1 = Cluster 1:  The active users, spend alot of time on the app, have high features usage, make use of alot of notification and so on
User 2 = Cluster 0: The medium users. They use the app not too much and also not too frequently. They have medium time spent and low features usage and so on.
User 3 = Cluster 2: The lowest users. They spend less, lower the Cluster 0 time on the app, they have the lowest features usage and so on
